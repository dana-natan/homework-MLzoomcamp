```python
asian_cars = df[df['origin'] == 'Asia']

X_df = asian_cars[['vehicle_weight', 'model_year']]

X_df.head(7)
```

```python
X = X_df.head(7).to_numpy()

X
```

```python
X.shape
```

```python
XTX = X.T @ X

XTX
```

```python
import numpy as np

XTX_inv = np.linalg.inv(XTX)

XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y

w
```
