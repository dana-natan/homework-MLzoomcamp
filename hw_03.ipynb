#### Handling missing values

The HW 3 instructions specify:
- Categorical missing values → "NA"
- Numerical missing values → 0.0

```python
df.dtypes # which type is each column
```

```python
categorical_columns = [
    'lead_source',
    'industry',
    'employment_status',
    'location'
]

df[categorical_columns] = df[categorical_columns].fillna("NA")

numerical_columns = [ #excluding the outcome variable
    'annual_income',
    'number_of_courses_viewed',
    'interaction_count',
    'lead_score'
]

df[numerical_columns] = df[numerical_columns].fillna(0)

df.isnull().sum() # n of missing in each column after filling
```

```python
from sklearn.model_selection import train_test_split

df_full_train, df_test = train_test_split(
    df, test_size=0.2, random_state=42
)

df_train, df_val = train_test_split(
    df_full_train, test_size=0.25, random_state=42 # the second split takes 25% of the remaining 80%, giving us 20% of the original dataset for validation.
)

df_train.shape, df_val.shape, df_test.shape
```

```python
y_train = df_train['converted']
X_train = df_train.drop(columns=['converted'])
```

## Question 3, Mutual Information

Mutual information measures how much knowing one variable reduces our uncertainty about another.

For example, if knowing a customer's lead_source tells us something useful about whether they'll convert, those variables have mutual information.

Unlike Pearson correlation, mutual information can capture nonlinear associations. It doesn't tell us the direction of the relationship.

The homework asks us to compare four categorical features: industry, location, lead_source, and employment_status.

```python
from sklearn.metrics import mutual_info_score
```

```python
mi = []
for column in categorical_columns:
    mi.append(mutual_info_score(X_train[column], y_train))

for column, sore in zip(categorical_columns, mi):
    print(column, round(sore, 2))
```

## Q4: Encode the training data

```python
train_dicts = X_train.to_dict(orient="records") # This converts each customer into a Python dictionary

train_dicts[0] # This is the first customer in the training set
```

```python
from sklearn.feature_extraction import DictVectorizer
dv = DictVectorizer(sparse=False) # save a DictVectorizer
```

```python
X_train_encoded = dv.fit_transform(train_dicts)
X_train_encoded.shape
```

- 3,000 rows: The same 3,000 customers from your training set.
- 28 columns: Your original 8 predictors have expanded into 28 numerical features because categorical variables now have separate indicator columns.
Notice that one-hot encoding doesn't create more observations. It creates more features.

```python
dv.get_feature_names_out() # This shows the 28 feature names that correspond to the columns in X_train_encoded.
# inspect them before training a model
```

**Create a Model**

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    solver='liblinear',
    C=1.0,
    max_iter=1000,
    random_state=42
)
```

The .fit() method **trains the model** by learning coefficients that help predict converted from the 28 numerical features:

```python
model.fit(X_train_encoded, y_train) # train the model
```

The model has now learned coefficients for the 28 features using the 3,000 training observations.

The next step is to prepare the validation dataset in the same way as we did for the training dataset. 
This involves separating the target variable from the features, converting the features into a list of dictionaries, 
and then transforming them using the DictVectorizer that was fitted on the training data. 

After that, we can use the trained model to make predictions on the validation set and evaluate its performance.

```python
X_val = df_val.drop(columns=['converted'])
y_val = df_val['converted']

val_dicts = X_val.to_dict(orient="records")
X_val_encoded = dv.transform(val_dicts) # use transform instead of fit_transform to avoid data leakage
```

```python
predictions = model.predict_proba(X_val_encoded)
y_pred = predictions[:, 1] >= 0.5

accuracy = np.mean(y_val == y_pred)
print(round(accuracy, 2))
```
